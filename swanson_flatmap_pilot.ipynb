# Download or Load Swanson Flatmap Atlas Data

```python
import numpy as np
from iblatlas.plots import plot_swanson_vector
from iblatlas.atlas import BrainRegions

br = BrainRegions()
```

# Inspect Atlas Metadata and Regions

```python
swanson_indices = np.unique(br.mappings['Swanson'])
swanson_ac = np.sort(br.acronym[swanson_indices])
print(f"Number of regions in Swanson flatmap: {swanson_ac.size}")
print("First 10 regions:", swanson_ac[:10])
```

# Visualize the Flatmap

```python
plot_swanson_vector(br=br, annotate=True)
```

# Map Sample Coordinates to Atlas Regions

```python
# Example: Plot values for sample regions
acronyms = ['VISa', 'CA1', 'VPM']
values = np.array([1.5, 3, 4])
plot_swanson_vector(acronyms, values, annotate=True, annotate_list=acronyms, empty_color='silver')
```
